```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# comapre different tanh functions with different  normalizations
x = np.linspace(-10, 10, 100)
d = [0.1, 0.5, 1.0, 2.0, 5.0]

# plot in a loop
for di in d:
    y1 = (1 - np.tanh(np.abs(x)/di))
    plt.plot(x, y1, label=f'tanh/ {di}')

plt.grid()
plt.legend()

# compare with just tanh(x) and tanh(x/10)
y2 = np.tanh(x)
y3 = np.tanh(x/10)
plt.plot(x, y2, label='tanh')
plt.plot(x, y3, label='tanh/10')
plt.grid()
plt.legend()
```

```python
import numpy as np
A = np.array([[-3, -7, -5],
              [1, 0, 0],
              [0, 1, 0]])
B = np.array([[1],
              [0],
              [0]])
```

```python
# determinant
np.linalg.det(C)
```
